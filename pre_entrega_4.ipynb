```python
%%writefile .env
API_URL=https://rickandmortyapi.com/api/character
#Este comando crea un archivo oculto llamado .env en tu entorno de Colab con la URL guardada de forma segura,
evitando el "hardcoding"
```

```python
import os
import requests
from dotenv import load_dotenv

# 1. Cargamos las variables de entorno desde el archivo .env
load_dotenv()

# 2. Obtenemos la URL de manera segura
url_api = os.getenv("API_URL")

print(f"📡 Conectándose a la API: {url_api}...\n")

try:
    # 3. Petición GET con control de tiempo de espera (timeout)
    respuesta = requests.get(url_api, timeout=10)

    # 4. Validamos códigos de error HTTP (como un 404 o 500)
    respuesta.raise_for_status()

    # Convertimos la respuesta a formato JSON
    data_json = respuesta.json()

    # Extraemos la lista de personajes que viene dentro de la clave 'results'
    personajes = data_json.get("results", [])

    if not personajes:
        print("⚠️ La API respondió correctamente, pero no devolvió personajes.")
    else:
        print(f"¡Conexión exitosa! Se recuperaron {len(personajes)} registros.")
        print("\nEjemplo del primer personaje obtenido:")
        print(f" - Nombre: {personajes[0].get('name')}")
        print(f" - Especie: {personajes[0].get('species')}")
        print(f" - Estado: {personajes[0].get('status')}")

except requests.exceptions.HTTPError as err_http:
    print(f"❌ Error HTTP: El servidor respondió con un código de error. Detalle: {err_http}")
except requests.exceptions.ConnectionError:
    print("❌ Error de Conexión: Verifica tu acceso a internet o si la URL es correcta.")
except requests.exceptions.Timeout:
    print("❌ Error: La petición tardó demasiado tiempo en responder (Timeout).")
except Exception as e:
    print(f"❌ Ocurrió un error inesperado: {e}")
```

```python
import pandas as pd

#Transformación Básica de los Datos
# Verificamos que la variable 'personajes' exista en la memoria
if 'personajes' in locals() and personajes:

    # Convertimos la lista de diccionarios en una base de datos
    df_personajes = pd.DataFrame(personajes)

    #Definimos qué columnas específicas nos interesan
    columnas_deseadas = ['id', 'name', 'status', 'species', 'gender']

    #Filtramos para conservar las columnas que nos importan de forma segura
    df_limpio = df_personajes[[col for col in columnas_deseadas if col in df_personajes.columns]].copy()

    print("¡Transformación completada con éxito! Las columnas innecesarias fueron sacadas.")

    #Mostramos una vista previa de las primeras filas para verificar visualmente el resultado
    display(df_limpio.head())

else:
    print("Advertencia: No se encontraron datos previos. Ejecuta primero la celda de extracción.")
```

```python
#Guardar los datos en un archivo estructurado CSV
# vemos que bd limpia ya haya sido creado en el paso anterior
if 'df_limpio' in locals() and not df_limpio.empty:

    #Asignamos el nombre obligatorio que pide el ejercisio
    nombre_archivo = "data_extracted.csv"

    #Exportamos los datos a un archivo CSV físico.
    #Usamos index=False para que no se guarde una columna extra con los números de fila automáticos.
    df_limpio.to_csv(nombre_archivo, index=False)

    print(f"¡Guardado exitoso! El archivo fue guardado como '{nombre_archivo}' en el entorno.")
    print("Acordate de descargarlo de la carpeta lateral para subirlo a tu Github y cumplir con los profes.")

else:
    print("Advertencia: No hay una base de datos limpia disponible. Asegúrate de ejecutar el paso anterior primero.")
```
